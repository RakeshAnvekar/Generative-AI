```python
## Dictionaries store the data in key value pairs, key muct be unique and immutable and it can include numbers,string or tuples
## Creation of Dictiories

empty_dict={}
print(empty_dict)
print(type(empty_dict))
```

```python
## accessing Dictionary elemnts
students={"name":"Rakesh","Age":31}

print(students["name"]) ##Rakesh
print(type(students["Age"])) ##<class 'int'>

# accesing using get() method of the dictionary

print(students.get("name")) ##Rakesh

print(students.get("name_test")) ##None

print(students.get("name_test"),"Not Available") ##None Not Available
```

```python
## modifying dictionary elements
## Dictionary are mutable,  we can update or delete elements

teachers={"name":"Rakesh","age":45,"type":"P"}
print(teachers) ##{'name': 'Rakesh', 'age': 45, 'type': 'P'}

## update
teachers["name"]="Rohan"
print(teachers) ##{'name': 'Rohan', 'age': 45, 'type': 'P'}

## adding new key and value
teachers["Address"]="Bangalore"
print(teachers)  ##{'name': 'Rohan', 'age': 45, 'type': 'P', 'Address': 'Bangalore'}

##delete the key
del teachers["Address"]
print(teachers) ## {'name': 'Rohan', 'age': 45, 'type': 'P'}
```

```python
## imp Dictioray method

## to get all the keys
print(teachers.keys()) ## dict_keys(['name', 'age', 'type'])

## to get values
print(teachers.values()) ## dict_values(['Rohan', 45, 'P'])

## get all the key value pairs

items=teachers.items() ## list of tuples
print(items) ##dict_items([('name', 'Rohan'), ('age', 45), ('type', 'P')])
```

```python
##shallow Copy
 ## issue
cites={"name":"bangalore","pin":1234}

print(cites)##{'name': 'bangalore', 'pin': 1234}

## copy to variable
cites_copy=cites

## change the value in dictionary
cites["name"]="hydrabad"
print(cites) ## {'name': 'hydrabad', 'pin': 1234}
print(cites_copy)## {'name': 'hydrabad', 'pin': 1234}

## Note: we have not made any changes to cites_copy but still it got changed
```

```python
## solution for the above problem

cites={"name":"bangalore","pin":1234}

print(cites)##{'name': 'bangalore', 'pin': 1234}

## copy to variable
cites_copy=cites.copy() ## shallow copy

## change the value in dictionary
cites["name"]="hydrabad"
print(cites) ## {'name': 'hydrabad', 'pin': 1234}
print(cites_copy)## {'name': 'bangalore', 'pin': 1234}
```

```python
## Iterating over Dictionary

## we can use loops to iterate over dict,keys,value or items

names={"name":"bangalore","pin":1234}
for key in names.keys():
    print(key)

for value in names.values():
    print(value)

for item in names.items():
    print(item)

for key,value in names.items():
    print(f"{key}:{value}")
```

```python
## nested Dictiories

my_Dict={
    "studen1":{"name":"rakesh","age":32},
    "studen2":{"name":"Rohan","age":30},
}

print(my_Dict)##{'studen1': {'name': 'rakesh', 'age': 32}, 'studen2': {'name': 'Rohan', 'age': 30}}

## access nested dictionary

print(my_Dict["studen1"]["age"]) ##32
```

```python
## iterating over nested dict


for dict_id,dict_info in my_Dict.items():
    print(f"{dict_id}:{dict_info}")
    for key,value in dict_info.items():
        print(f"{key}:{value}")
    
```

```python
## dictionary comprehension
## what is the square of the gien elements in dict

numbers={x:x*2  for x in range(5)}
print(numbers) ##{0: 0, 1: 2, 2: 4, 3: 6, 4: 8}
```

```python
## what is the square of the  even elements in dict

even_numbers_square={x:x*2 for x in range(5) if x %2==0}
print(even_numbers_square) ##{0: 0, 2: 4, 4: 8}
```

```python
##Practicle examples
##use the dict to count the frequency of the elements in the list

nubers=[1,2,3,3,4,2,4,4,4]

frequency={}
for number in nubers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## merge two dictionary into one
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}

merge_dict={**dict1,**dict2}
print(merge_dict) ## {'a': 1, 'b': 3, 'c': 4}
```
