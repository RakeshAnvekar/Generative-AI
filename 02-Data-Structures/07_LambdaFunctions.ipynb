```python
##syntax
## lambda arguments:expression

def addition(a,b):
    return a+b

addition(2,3)


addition= lambda a,b:a+b
print(addition(1,2))


even= lambda n:n%2==0
even(2)
```

```python
def addtion(n1,n2,n3):
    return n1+n2+n3

addtion(1,2,3)

add_lamba=lambda num1,num2,num3:num1+num2+num3

print(add_lamba(1,2,4))
```

```python
## map(), appplies a function to all items in the list

numbers=[1,2,3,4]

## find the square of each and every number
list(map(lambda x:x**2,numbers))
```
