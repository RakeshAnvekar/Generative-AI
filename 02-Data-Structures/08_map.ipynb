```python
def squere(n):
    return n*n

# with labda and map to apply same for all elements list

numbers=[1,2,3,4,5]

list(map(lambda n:n*n,numbers))
```

```python
## function with map

def square(n):
    return n*n

numbers=[2,4,8]

list(map(square,numbers))
```

```python
## map multipe iterables
numbers1=[1,2,3]
numbers2=[4,5,6]

list(map(lambda a,b:a+b,numbers1,numbers2))
```

```python
##map to convert list of string to int
str_numbers=["1","2","3","4"]

list(map(int,str_numbers))
```

```python
words=['apple','banana','orange']
list(map(str.upper,words))
```

```python
## apply map to list of dictionary
def get_name(person):
    return person["name"]

people=[
    {'name':'rakesh','age':32},
    {'name':'vinod','age':62},
]

list(map(get_name,people))
```
