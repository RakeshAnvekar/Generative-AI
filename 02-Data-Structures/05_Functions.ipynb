```python
## varaiable length arguments
## positional and keywords argument
def print_numbers(*args): ## insted of having the metod declartion as print_numbers(a,b,c,d,e)
    for number in args:
        print(number)

print_numbers(1,2,3,4,"r")
```

```python
## keywords arguments - all the parameters will be in key and value pair
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"key : {key} -- value :{value}")

print_details(name="rakesh",city="bangalore",contry="india")
```

```python
def print_positonal_and_keywords(*args,**kwargs):
    for val in args:
        print(f"args- {val}")

    for key,value in kwargs.items():
            print(f"kwargs- key : {key} -- value :{value}")

print_positonal_and_keywords(1,2,3,4,name="rakesh",city="bangalore",contry="india")
```
