```python
def even(num):
    if num%2==0:
        return True
## filer function filter the items from the iteratable and filter based on the conditon

lists=[1,2,3,4,5,7,9]
list(filter(even,lists))
## or
```

```python
## filyer with labda function

list(filter(lambda x:x%2==0,lists))

numbers=[4,7,5,1,10,11,6,7]
list(filter(lambda x:x>5,numbers))
```

```python
## filter with multiple condition and lambda function
numbers=[4,7,5,1,10,11,6,7]
list(filter(lambda x:x>5 and x%2==0,numbers)) 


```

```python
## filter operation on dictionary
## to check age is greater then 25 in dictionary

peoples=[

    {'name':'rakesh','age':26},
    {'name':'Vinod','age':20},
]

list(filter(lambda people: int(people["age"])> 25 ,peoples))

```
