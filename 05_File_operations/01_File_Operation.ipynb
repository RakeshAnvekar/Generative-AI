```python
## Read a whole file

with open("example.txt",'r') as file:
    content=file.read()
    print(content)
```

```python
## read file line by line
with open("example.txt",'r') as file:
    for line in file:
        print(line.strip()) ##.strip() removes the new line char
```

```python
##open a file for writing a file

with open("example.txt",'w') as file:
    file.write("New Data\n")
    file.write('this ia new line\n') ## Entire content got overridden
    
```

```python
## write a file without overriding the content
with open('example.txt','a') as file:
    file.write("new one\n")
```

```python
## write multiple lines to a file
lines=['Fist Line\n','Second line\n','third line\n']
with open('example.txt','a') as file:
    file.writelines(lines) 
```

```python
## Binary files -example_binary.bin
## writing to a binary file
data =b'\x00\x01\x02\x03'
with open('example_binary.bin','wb') as binaryFile:
    binaryFile.write(data)
```

```python
## reding from the binary files
with open('example_binary.bin','rb') as binaryFile:
    print(binaryFile.read())
```

```python
## read text file and coun the number of lines words and char

def count_text_file(filePath):
    with open(filePath, 'r') as file:
        file_lines = file.readlines()

        line_count = len(file_lines)
        word_count = sum(len(line.split(" ")) for line in file_lines)

    return line_count, word_count


filePath = "example.txt"

lines, word_count = count_text_file(filePath)

print(f"Lines: {lines}")
print(f"Words: {word_count}")
```

```python
## writing and then reading the file
## w+ mode in the python is used to open the file for both reading and writing,if the file doesnot exist it will be created,
with open('example.txt','w+') as file:
    file.write("hello world\n")
    file.write("this is new line\n")

    ## move the file cursor to first so we can read it

    file.seek(0)

    print(file.read())
```

```python
# ways to check the file is present or not
path="example.txt"
if os.path.exists(path):
    print(f"{path} exists")
else:
    print(f"{path}  does not exists")
```

```python
## Checking if path is a file or directory

path="example.txt"
if os.path.isfile(path):
    print(f"The path  {path}  is a file")
elif os.path.isdir(path):
    print(f"The pah {path}  is a directory")
else:
    print(f"the path {path}  is nither a file or a directory")
```
