```python
## try,except ,else block

try:
    num=int(input("Enter the number"))
    result=10/num
except ValueError as ex:
    print(ex)
except ZeroDivisionError as ex:
    print(ex)
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}") 
    ## else block is executed if you want to run som code if no exceotion is raised, 
    ## else block will  not executed if we encounter any exception, 
    ## we no exception the else blcok will get executed
```

```python
## try,except else and finally



try:
    num=int(input("Enter he number"))
    result=10/num
except ZeroDivisionError as ex:
    print(ex)
except ValueError as ex:
    print(ex)
except Exception as ex:
    print(ex)
else:
    print(F"This is else blcok it executes if no excetion occur {result}")
finally:
    print("This code always executes")
```

```python
## file handlig and exception handling
try:
    file=open("example.txt",'r')
    content= file.read()
    print(content)
except FileNotFoundError as ex:
    print(ex)
except Exception as ex:
    print(ex)
finally:
   if 'file' in locals() and not file.closed:
        print("File closed")
        file.close()
        

```
